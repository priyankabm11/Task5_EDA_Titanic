```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
sns.countplot(x='survived', data=titanic)
plt.title("Survival Count")
plt.show()
```

```python
sns.histplot(titanic['age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(x=titanic['fare'])
plt.title("Fare Distribution")
plt.show()
```

```python
sns.countplot(x='sex', hue='survived', data=titanic)
plt.title("Survival by Gender")
plt.show()
```

```python
sns.scatterplot(
    x='age',
    y='fare',
    hue='survived',
    data=titanic
)
plt.title("Age vs Fare")
plt.show()
```

```python
corr = titanic.select_dtypes(include='number').corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(
    titanic[['age','fare','pclass','survived']].dropna()
)
plt.show()
```

# Summary of Findings

1. Female passengers had a higher survival rate than males.
2. First-class passengers survived more often.
3. Fare shows a positive relationship with survival.
4. Age has a weak effect on survival.
5. Fare contains several outliers.
6. Missing values are present in age and deck columns.
